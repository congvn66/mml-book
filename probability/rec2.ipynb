## 1.

$A$: coin 1 is head.

$B$: coin 2 is head.

We can now phrase Alice's claim as: $P(AB|A) \ge P(AB|(A\cup B))$

$$P(AB|A) = \frac{P(AB∩A)}{P(A)} = \frac{P(AB)}{P(A)}$$

$$P(AB|(A\cup B)) = \frac{P(AB \cap (A\cup B))}{P(A\cup B)} = \frac{P(AB)}{P(A\cup B)}$$

Observe that, $P(A) \le P(A \cup B) ∀ A, B$. Thus, Alice's claim is correct, with no difference if the coin is fair or not.

We can generalize the situation as below:

$$C, D, E$$
$$D ⊆ E$$
$$C \cap D = C \cap E$$
$$⇒ P(C \cap D | D) \ge P(C \cap E | E)$$
